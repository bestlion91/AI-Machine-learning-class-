```python
ikkilantir(5)
```

```python
# berilgan istalgan sonlarni yigindisini topadigan 
# berilgan listdagi so'zlar ichidan eng qisqa so'zni(birinchi keladigan) toping
# berilgan listdagi so'zlar ichidan eng uzun so'zni(birinchi keladigan) toping
# berilgan sonlar ichidan juftlarini topadigan 
# berilgan sonlar ichidan toqlarini
# berilgan sonlar ichidan juft/toqlarini yigindisini topadigan funksiya yarating
```

```python
# berilgan istalgan sonlarni yigindisini topadigan funksiya tuzing;
def umumiy_yigindi(*sonlar):
    return sum(sonlar)
```

```python
umumiy_yigindi(25,6,35)
```

```python
# berilgan listdagi so'zlar ichidan eng qisqa so'zni(birinchi keladigan) toping;
matn='I am tall, about six feet. I have longish brown hair. My eyes are green and I have a wide mouth'
def find_short_word(matn):
    soz=matn.split()
    soz.sort(key=len)
    return soz[0]
```

```python
find_short_word(matn)
```

```python
# berilgan listdagi so'zlar ichidan eng uzun so'zni(birinchi keladigan) toping
matn='I am tall, about six feet. I have longish brown hair. My eyes are green and I have a wide mouth'
def find_short_word(matn):
    soz=matn.split()
    soz.sort(key=len)
    return soz[-1]
```

```python
find_short_word(matn)
```

```python
# berilgan sonlar ichidan juftlarini topadigan funksiya tuzing;
def juftlarini_top(*sonlar):
    juftlar = []
    for i in sonlar:
        if i % 2 == 0:
            juftlar.append(i)
    return juftlar
```

```python
juftlarini_top(25,35,48,69,477,2563,4,8)
```

```python
# berilgan sonlar ichidan toqlarini topadigan funksiya tuzing;
def juftlarini_top(*sonlar):
    juftlar = []
    for i in sonlar:
        if i%2!=0:
            juftlar.append(i)
    return juftlar
```

```python
juftlarini_top(25,35,48,69,477,2563,4,8)
```

```python
def juft_va_toqlarni_top(*sonlar):
    juftlar = []
    toqlar = []
    for i in sonlar:
        if i % 2 == 0:
            juftlar.append(i)
        elif i % 2 !=0:
            toqlar.append(i)
    print(f"juft sonlar quyidagilardir:{juftlar}")
    print(f"toq sonlar quyidagilardir:{toqlar}")
    return i
```

```python
juft_va_toqlarni_top(25,35,48,69,477,2563,4,8)
```

```python
def juft_va_toqlarni_top(*sonlar):
    juftlar = []
    toqlar = []
    for i in sonlar:
        if i % 2 == 0:
            juftlar.append(i)
        else:
            toqlar.append(i)
    print(f"juft sonlar quyidagilardir:{juftlar}")
    print(f"toq sonlar quyidagilardir:{toqlar}")
    return i
```

```python
# berilgan sonlar ichidan juft/toqlarini yigindisini topadigan funksiya yarating;
def juft_va_toqlarni_top(*sonlar):
    juftlar = []
    toqlar = []
    for i in sonlar:
        if i % 2 == 0:
            juftlar.append(i)
        else:
            toqlar.append(i)     
    print(f"juft sonlar quyidagilardir:{juftlar}")
    print(f"toq sonlar quyidagilardir:{toqlar}")
    return sum(juftlar+toqlar)
```

```python
juft_va_toqlarni_top(25,35,48,69,477,2563,4,8)
```
