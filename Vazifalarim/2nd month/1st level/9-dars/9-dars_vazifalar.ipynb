```python
import pandas as pd
```

```python
df=pd.read_csv('movies_initial.csv')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().sum() > 10000:
        # 1) Agar NaN soni 10000 dan ko'p bo'lsa, ustunni o'chirish
        df.drop(columns=[col], inplace=True)
    elif df[col].dtype == 'object':
        # 2) Agar ustun turi object bo‘lsa va NaN soni 10 dan ko‘p bo‘lsa
        if df[col].isnull().sum() > 10:
            df[col].fillna(df[col].mode().iloc[0], inplace=True)
    else:
        # 3) Aks holda (sonli ustunlarda) NaN qiymatlarni o'rtacha bilan to'ldirish
        df[col].fillna(df[col].mean(), inplace=True)
```

```python
def encode_qil(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            encoder = LabelEncoder()
            df[col] = encoder.fit_transform(df[col].astype(str))
    return df
            
```
