```python
# Berilgan listdagi sonlarni tartiblang agar bosh bolsa yoki none bolsa [] ni qaytaring
def tartibla(nums):
    if nums is None or len(nums)==0:
        return []
   
    return sorted(nums)



```

```python
tartibla([6,3,2])
```

# 1

```python
#1 Robotni turgan  positisiyasi  berilgan va robotni ishlash qoidasiga kora chapdan onga xarakatlanadi va berilgan buyruq songa ko'ra 3 marta xarakatlanadi va bir marta orqaga xarakatlanadi. Robotni yangi joyini topadigan funksiya yarating:
# Masalan  start=0 buyruq=2 natija 4
def joyni_top(a): return f"Robotning yangi joyi bu yerda: {(a*3)-a}"
```

```python
joyni_top(6)
```

# 2

```python
#2 Berilgan  2 ta kubning orasidan eng kattasini toping(1 qator kod orqali)
def kattasini_top(a, b): return f"{a} ning kubi {b} ning kubidan katta" if a**3 > b**3 else f"{b} ning kubi {a} ning kubidan katta"
```

```python
kattasini_top(25,98)
```

```python
# 4 Berilgan  2 ta kubning  hajmi orasidagi farqni toping
def faqni_top(a, b): return f"{a} sonning kubi bilan {b} sonning kubi orasidagi farq quyidagicha: {a**3 - b**3}"
```

```python
faqni_top(30,215)
```
