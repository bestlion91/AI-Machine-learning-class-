# Funksiyalarni bitta classga birlashtiramiz

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
import pandas as pd

class Datapreprocessing:
    def __init__(self, df):
        self.df = df.copy()
        self.encoder = LabelEncoder()
        self.scaler = MinMaxScaler()   

    def tozala(self):
        for col in self.df.columns:
            if self.df[col].isnull().any():
                if self.df[col].dtype == 'object':
                    self.df[col] = self.df[col].fillna(self.df[col].mode()[0])
                else:
                    self.df[col] = self.df[col].fillna(self.df[col].mean())
        return self

    def encodla(self):
        for col in self.df.columns:
            if self.df[col].dtype == 'object':
                if self.df[col].nunique() <= 5:   # categoriyalar kam bo‘lsa One-Hot
                    dummies = pd.get_dummies(self.df[col], prefix=col, dtype=int)
                    self.df = pd.concat([self.df.drop(columns=[col]), dummies], axis=1)
                else:   
                    self.df[col] = self.encoder.fit_transform(self.df[col])
        return self

    def scale_qil(self):
        num_cols = self.df.select_dtypes(include=['int64', 'float64']).columns
        self.df[num_cols] = self.scaler.fit_transform(self.df[num_cols])
        return self
```

```python
processor = Datapreprocessing(df)

df_ready = (
    processor
    .tozala()
    .encodla()
    .scale_qil()
    .df
)
print(df_ready)
```

# Model Training

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop("mpg", axis=1)   # bular feautures
y = df["mpg"] # bu esa target qiymat qilib belgilab oldim
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42) # train test split yordamida qismlarga ajratib olamiz
```

```python
x_train.shape
```

```python
x_test.shape
```

# Model Selection

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

# Linear Regression

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
lr_mse = mean_squared_error(y_test, y_pred)
lr_mae = mean_absolute_error(y_test, y_pred)
lr_r2 = r2_score(y_test, y_pred)

print("MSE:", lr_mse)
print("MAE:", lr_mae)
print("R2 score:", lr_r2)
```

# Random Forest Regressor

```python
rf = RandomForestRegressor(random_state=42)
rf.fit(x_train, y_train)
```

```python
y_pred = rf.predict(x_test)
```

```python
rf_mse = mean_squared_error(y_test, y_pred)
rf_mae = mean_absolute_error(y_test, y_pred)
rf_r2 = r2_score(y_test, y_pred)

print("MSE:", rf_mse)
print("MAE:", rf_mae)
print("R2 score:", rf_r2)
```

# DecisionTreeRegressor

```python
dt = DecisionTreeRegressor()
dt.fit(x_train, y_train)
```

```python
y_pred = dt.predict(x_test)
```

```python
dt_mse = mean_squared_error(y_test, y_pred)
dt_mae = mean_absolute_error(y_test, y_pred)
dt_r2 = r2_score(y_test, y_pred)

print("MSE:", dt_mse)
print("MAE:", dt_mae)
print("R2 score:", dt_r2)
```
